## 1.9 File I/O

```python
import tempfile
import os

# Write to file in temp directory
temp_file = os.path.join(tempfile.gettempdir(), "example.txt")
with open(temp_file, "w") as f:
    f.write("Hello, World!\n")
    f.write("Python is awesome!\n")

print(f"File written to {temp_file}")
```

```python
# Read from file
with open(temp_file, "r") as f:
    content = f.read()
print(content)
```

```python
# JSON files
import json

json_file = os.path.join(tempfile.gettempdir(), "data.json")
data = {"name": "Alice", "age": 30, "city": "NYC"}

# Write JSON
with open(json_file, "w") as f:
    json.dump(data, f)

# Read JSON
with open(json_file, "r") as f:
    loaded_data = json.load(f)

print(loaded_data)
```

## 1.10 Decorators

```python
def my_decorator(func):
    def wrapper():
        print("Before function call")
        func()
        print("After function call")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
```

```python
import time

def timing_decorator(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Execution time: {end - start:.4f}s")
        return result
    return wrapper

@timing_decorator
def slow_function():
    time.sleep(0.5)
    return "Done"

result = slow_function()
```

## 1.11 Generators

```python
def count_up(n):
    i = 1
    while i <= n:
        yield i
        i += 1

for num in count_up(5):
    print(num)
```

```python
# Generator expression
gen = (x*2 for x in range(5))
print(list(gen))
```

## 1.12 Iterators

```python
class CountUp:
    def __init__(self, max):
        self.max = max
        self.current = 0
    
    def __iter__(self):
        return self
    
    def __next__(self):
        self.current += 1
        if self.current <= self.max:
            return self.current
        else:
            raise StopIteration

counter = CountUp(3)
for num in counter:
    print(num)
```

## 1.13 Type Hints

```python
def add(a: int, b: int) -> int:
    return a + b

result = add(5, 3)
print(f"Result: {result}")
```

```python
from typing import List, Dict, Optional

def process_items(items: List[str]) -> Dict[str, int]:
    return {item: len(item) for item in items}

result = process_items(["apple", "banana", "cherry"])
print(result)
```

## 1.14 Dataclasses

```python
from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int
    email: str

person = Person("Alice", 25, "alice@example.com")
print(person)
print(f"Name: {person.name}")
```

```python
from dataclasses import dataclass, field

@dataclass
class Config:
    host: str
    port: int = 8080
    debug: bool = False
    tags: list = field(default_factory=list)

config1 = Config("localhost")
config2 = Config("example.com", 3000, True, ["prod"])

print(config1)
print(config2)
```

## 1.15 Context Managers

```python
# File auto-closes
test_file = os.path.join(tempfile.gettempdir(), "test.txt")
with open(test_file, "w") as f:
    f.write("Hello!")
    print(f"File is open: {not f.closed}")

print(f"File is closed: {f.closed}")
```

```python
class FileManager:
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode
        self.file = None
    
    def __enter__(self):
        self.file = open(self.filename, self.mode)
        print(f"Opened: {self.filename}")
        return self.file
    
    def __exit__(self, exc_type, exc_val, exc_tb):
        if self.file:
            self.file.close()
        print(f"Closed: {self.filename}")

custom_file = os.path.join(tempfile.gettempdir(), "custom.txt")
with FileManager(custom_file, "w") as f:
    f.write("Custom context manager!")
```
