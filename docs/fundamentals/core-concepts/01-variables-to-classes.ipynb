## 1.1 Variables

Variables are named storage locations that hold data values. Python uses dynamic typing.

```python
# Variable assignment
x = 5
name = "Alice"
price = 19.99

print(f"x = {x}, type = {type(x)}")
print(f"name = {name}, type = {type(name)}")
print(f"price = {price}, type = {type(price)}")
```

```python
# Dynamic typing - type changes with assignment
x = 5
print(f"x = {x}, type = {type(x).__name__}")

x = "hello"
print(f"x = {x}, type = {type(x).__name__}")

x = 3.14
print(f"x = {x}, type = {type(x).__name__}")
```

## 1.2 Data Types

Python has built-in data types: int, str, float, bool, None, list, dict, tuple, set.

```python
# Basic data types
integer = 42
floating = 3.14
string = "Python"
boolean = True
none_value = None

print(f"int: {integer}")
print(f"float: {floating}")
print(f"str: {string}")
print(f"bool: {boolean}")
print(f"None: {none_value}")
```

```python
# Collections
my_list = [1, 2, 3, "four", 5.0]
my_tuple = (1, 2, 3)
my_dict = {"name": "Alice", "age": 30}
my_set = {1, 2, 3, 2, 1}  # Duplicates removed

print(f"list: {my_list}")
print(f"tuple: {my_tuple}")
print(f"dict: {my_dict}")
print(f"set: {my_set}")
```

## 1.3 Operators

Arithmetic, comparison, logical, and assignment operators.

```python
# Arithmetic operators
a, b = 10, 3

print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}")
print(f"{a} // {b} = {a // b}")
print(f"{a} % {b} = {a % b}")
print(f"{a} ** {b} = {a ** b}")
```

```python
# Comparison operators
x, y = 5, 10

print(f"{x} > {y}: {x > y}")
print(f"{x} < {y}: {x < y}")
print(f"{x} == {y}: {x == y}")
print(f"{x} != {y}: {x != y}")
print(f"{x} >= {y}: {x >= y}")
print(f"{x} <= {y}: {x <= y}")
```

```python
# Logical operators
print(f"True and False: {True and False}")
print(f"True or False: {True or False}")
print(f"not True: {not True}")

a, b = 5, 10
print(f"\n(a > 0) and (b > 0): {(a > 0) and (b > 0)}")
print(f"(a > 0) or (b < 0): {(a > 0) or (b < 0)}")
```

## 1.4 Control Flow

if/elif/else, for loops, while loops, break, continue.

```python
# if/elif/else
age = 20

if age < 13:
    print("Child")
elif age < 18:
    print("Teenager")
elif age < 65:
    print("Adult")
else:
    print("Senior")
```

```python
# for loop
print("For loop:")
for i in range(5):
    print(f"i = {i}")

print("\nFor loop with list:")
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"  {fruit}")
```

```python
# while loop
count = 0
while count < 3:
    print(f"count = {count}")
    count += 1

print("\nLoop with break:")
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
# continue
print("Loop with continue (skip even numbers):")
for i in range(6):
    if i % 2 == 0:
        continue
    print(i)
```

## 1.5 Functions

Define reusable code blocks with parameters and return values.

```python
# Simple function
def greet(name):
    return f"Hello, {name}!"

result = greet("Alice")
print(result)
```

```python
# Function with multiple parameters and return
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

print(f"add(3, 5) = {add(3, 5)}")
print(f"multiply(3, 5) = {multiply(3, 5)}")
```

```python
# Function with default parameters
def power(base, exponent=2):
    return base ** exponent

print(f"power(5) = {power(5)}")
print(f"power(5, 3) = {power(5, 3)}")
```

```python
# Recursion
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(f"factorial(5) = {factorial(5)}")
print(f"factorial(10) = {factorial(10)}")
```

## 1.6 Classes & Objects

Create reusable blueprints for objects with attributes and methods.

```python
# Simple class
class Dog:
    species = "Canis familiaris"
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def bark(self):
        return f"{self.name} says: Woof!"
    
    def get_age(self):
        return self.age

# Create instances
dog1 = Dog("Rex", 3)
dog2 = Dog("Buddy", 5)

print(f"dog1 name: {dog1.name}")
print(f"dog1 species: {dog1.species}")
print(f"dog1.bark(): {dog1.bark()}")
print(f"dog2.get_age(): {dog2.get_age()}")
```

```python
# Class with encapsulation
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance  # Protected
    
    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            return f"Deposited ${amount}"
        return "Invalid amount"
    
    def withdraw(self, amount):
        if 0 < amount <= self._balance:
            self._balance -= amount
            return f"Withdrew ${amount}"
        return "Insufficient funds"
    
    def check_balance(self):
        return self._balance

# Usage
account = BankAccount("Alice", 1000)
print(account.deposit(500))
print(account.withdraw(200))
print(f"Balance: ${account.check_balance()}")
```

```python
# Class methods and static methods
class Calculator:
    def __init__(self):
        self.result = 0
    
    def add(self, x):
        self.result += x
        return self.result
    
    @classmethod
    def from_value(cls, initial_value):
        calc = cls()
        calc.result = initial_value
        return calc
    
    @staticmethod
    def multiply(a, b):
        return a * b
    
    def __str__(self):
        return f"Calculator(result={self.result})"

# Usage
calc = Calculator()
print(calc.add(5))
print(calc.add(10))

calc2 = Calculator.from_value(100)
print(calc2)

print(f"multiply(3, 4) = {Calculator.multiply(3, 4)}")
```

## Mini-Exercise: Rectangle Class

Create a Rectangle class with area() and perimeter() methods.

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)
    
    def __str__(self):
        return f"Rectangle({self.width}x{self.height})"

rect = Rectangle(5, 3)
print(rect)
print(f"Area: {rect.area()}")
print(f"Perimeter: {rect.perimeter()}")
```
