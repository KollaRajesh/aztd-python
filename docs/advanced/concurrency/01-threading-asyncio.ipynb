## 2.7-2.8 Threading

```python
import threading
import time

def worker(name, delay):
    print(f"{name} starting")
    time.sleep(delay)
    print(f"{name} done")

# Create threads
threads = [
    threading.Thread(target=worker, args=("Thread 1", 1)),
    threading.Thread(target=worker, args=("Thread 2", 2)),
    threading.Thread(target=worker, args=("Thread 3", 1))
]

# Start threads
for thread in threads:
    thread.start()

# Wait for completion
for thread in threads:
    thread.join()

print("All threads completed")
```

```python
import threading

# Thread synchronization with Lock
counter = 0
lock = threading.Lock()

def increment():
    global counter
    for _ in range(100000):
        with lock:
            counter += 1

threads = [threading.Thread(target=increment) for _ in range(5)]

for t in threads:
    t.start()

for t in threads:
    t.join()

print(f"Final counter: {counter}")
```

## 2.9-2.10 Asyncio & async/await

```python
import asyncio

async def fetch_data(name, delay):
    print(f"{name} fetching...")
    await asyncio.sleep(delay)
    print(f"{name} done")
    return f"Data from {name}"

async def main():
    # Run concurrent tasks
    tasks = [
        fetch_data("API 1", 1),
        fetch_data("API 2", 2),
        fetch_data("API 3", 1)
    ]
    results = await asyncio.gather(*tasks)
    print(f"Results: {results}")

# Run async main
await main()
```

```python
import asyncio

async def producer(queue):
    for i in range(5):
        await queue.put(i)
        print(f"Produced: {i}")
        await asyncio.sleep(0.1)
    await queue.put(None)  # Signal end

async def consumer(queue):
    while True:
        item = await queue.get()
        if item is None:
            break
        print(f"Consumed: {item}")

async def main():
    queue = asyncio.Queue()
    await asyncio.gather(
        producer(queue),
        consumer(queue)
    )

await main()
```

## 2.11 Threading vs Asyncio Comparison

```python
import time
import threading
import asyncio

# Simulate I/O operation
def io_operation():
    time.sleep(0.1)
    return "done"

# Threading approach
start = time.time()
threads = [threading.Thread(target=io_operation) for _ in range(10)]
for t in threads:
    t.start()
for t in threads:
    t.join()
threading_time = time.time() - start
print(f"Threading time: {threading_time:.3f}s")

# Asyncio approach
async def async_io():
    await asyncio.sleep(0.1)
    return "done"

async def main():
    tasks = [async_io() for _ in range(10)]
    await asyncio.gather(*tasks)

start = time.time()
await main()
async_time = time.time() - start
print(f"Asyncio time: {async_time:.3f}s")
```
