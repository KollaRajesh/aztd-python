## Simple: GET & POST

```python
import requests
import json

# GET request example
response = requests.get("https://jsonplaceholder.typicode.com/users/1")
print(f"Status: {response.status_code}")
print(json.dumps(response.json(), indent=2))
```

```python
# POST request example
payload = {"title": "Test Post", "body": "This is a test", "userId": 1}
response = requests.post("https://jsonplaceholder.typicode.com/posts", json=payload)
print(f"Status: {response.status_code}")
print(json.dumps(response.json(), indent=2))
```

## Medium: Headers, Params & Error Handling

```python
import requests

headers = {"User-Agent": "MyApp/1.0"}
params = {"_page": 1, "_limit": 5}

try:
    response = requests.get(
        "https://jsonplaceholder.typicode.com/posts",
        headers=headers,
        params=params,
        timeout=5
    )
    response.raise_for_status()
    posts = response.json()
    print(f"Retrieved {len(posts)} posts")
except requests.exceptions.RequestException as e:
    print(f"Error: {e}")
```

## Complex: Sessions & Retries

```python
import requests
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

def create_session():
    session = requests.Session()
    retry_strategy = Retry(
        total=3,
        backoff_factor=1,
        status_forcelist=[429, 500, 502, 503, 504]
    )
    adapter = HTTPAdapter(max_retries=retry_strategy)
    session.mount("http://", adapter)
    session.mount("https://", adapter)
    return session

session = create_session()
response = session.get("https://jsonplaceholder.typicode.com/users")
print(f"Retrieved {len(response.json())} users")
session.close()
```
