## Simple: Load .env

```python
from dotenv import load_dotenv
import os

# Create a sample .env file
with open('.env', 'w') as f:
    f.write('API_KEY=secret123\n')
    f.write('DEBUG=True\n')

load_dotenv()

API_KEY = os.getenv("API_KEY")
DEBUG = os.getenv("DEBUG", "False").lower() == "true"

print(f"API Key: {API_KEY}")
print(f"Debug: {DEBUG}")
```

## Medium: Type Conversion & Defaults

```python
from dotenv import load_dotenv
import os

# Create sample .env
with open('.env', 'w') as f:
    f.write('DATABASE_URL=sqlite:///app.db\n')
    f.write('PORT=8000\n')
    f.write('ALLOWED_HOSTS=localhost,127.0.0.1\n')
    f.write('MAX_CONNECTIONS=10\n')

load_dotenv()

DATABASE_URL = os.getenv("DATABASE_URL", "sqlite:///app.db")
PORT = int(os.getenv("PORT", "8000"))
ALLOWED_HOSTS = os.getenv("ALLOWED_HOSTS", "localhost").split(",")
MAX_CONNECTIONS = int(os.getenv("MAX_CONNECTIONS", "10"))

print(f"DB: {DATABASE_URL}")
print(f"Port: {PORT}")
print(f"Hosts: {ALLOWED_HOSTS}")
print(f"Max connections: {MAX_CONNECTIONS}")
```

## Complex: Config Class with Pydantic

```python
from dotenv import load_dotenv
import os
from pydantic import BaseModel

# Create sample .env
with open('.env', 'w') as f:
    f.write('API_KEY=my_secret_key\n')
    f.write('DATABASE_URL=postgresql://user:pass@localhost/db\n')
    f.write('DEBUG=false\n')
    f.write('PORT=8000\n')

load_dotenv()

class Settings(BaseModel):
    api_key: str = os.getenv("API_KEY", "")
    database_url: str = os.getenv("DATABASE_URL", "sqlite:///app.db")
    debug: bool = os.getenv("DEBUG", "false").lower() == "true"
    port: int = int(os.getenv("PORT", "8000"))

settings = Settings()
print(settings.model_dump())
```

## Usage in FastAPI App

```python
# Example integration with FastAPI
from fastapi import FastAPI
from dotenv import load_dotenv
import os

load_dotenv()

DEBUG = os.getenv("DEBUG", "false").lower() == "true"
PORT = int(os.getenv("PORT", "8000"))

app = FastAPI(debug=DEBUG)

@app.get("/config")
def get_config():
    return {"debug": DEBUG, "port": PORT}

print(f"FastAPI app configured with debug={DEBUG}, port={PORT}")
```
