## Simple: Parse HTML

```python
from bs4 import BeautifulSoup

html = "<html><body><p class='title'>Hello</p><p>World</p></body></html>"
soup = BeautifulSoup(html, "html.parser")

# Find single element
title = soup.find("p", class_="title")
print(title.text)
```

```python
# Find all
paragraphs = soup.find_all("p")
for p in paragraphs:
    print(p.text)
```

## Medium: Web Scraping

```python
# Example with local HTML
html_content = """
<html>
<body>
  <h2>Python Guide</h2>
  <h2>FastAPI Tutorial</h2>
  <a href="/page1">Page 1</a>
  <a href="/page2">Page 2</a>
  <div class="item">
    <p class="name">Item 1</p>
  </div>
</body>
</html>
"""

soup = BeautifulSoup(html_content, "html.parser")

# Extract data
titles = [h.text for h in soup.find_all("h2")]
print("Titles:", titles)

links = [a["href"] for a in soup.find_all("a")]
print("Links:", links)
```

```python
# CSS selectors
items = soup.select("div.item > p.name")
for item in items:
    print(item.text)
```

## Complex: Navigation & Error Handling

```python
def extract_data(html_string):
    try:
        soup = BeautifulSoup(html_string, "html.parser")
        
        # Extract title
        title = soup.title
        title_text = title.string if title else "No title"
        
        # Extract all paragraphs
        paragraphs = soup.find_all("p")
        para_texts = [p.get_text(strip=True) for p in paragraphs]
        
        # Extract links
        links = []
        for a in soup.find_all("a", href=True):
            links.append({"text": a.text, "url": a["href"]})
        
        return {
            "title": title_text,
            "paragraphs": para_texts,
            "links": links
        }
    except Exception as e:
        print(f"Error: {e}")
        return None

test_html = """
<html>
<head><title>Test Page</title></head>
<body>
  <p>First paragraph</p>
  <p>Second paragraph</p>
  <a href="/page1">Link 1</a>
</body>
</html>
"""

data = extract_data(test_html)
print(data)
```
