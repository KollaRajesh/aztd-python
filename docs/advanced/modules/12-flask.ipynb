## Simple: Basic App

````python
from flask import Flask, jsonify

app = Flask(__name__)

@app.route("/")
def home():
    return "Hello, Flask!"

@app.route("/api/users", methods=["GET"])
def get_users():
    return jsonify({"users": ["Alice", "Bob"]})

print("Flask app defined")
````

## Medium: Routing & Request Handling

````python
from flask import Flask, jsonify, request

app = Flask(__name__)

@app.route("/users/<int:user_id>")
def get_user(user_id):
    return jsonify({"id": user_id, "name": f"User {user_id}"})

@app.route("/create", methods=["POST"])
def create_user():
    data = request.json
    return jsonify({"status": "created", "user": data}), 201

@app.route("/users/<int:user_id>", methods=["PUT"])
def update_user(user_id):
    data = request.json
    return jsonify({"id": user_id, "updated": data})

@app.route("/users/<int:user_id>", methods=["DELETE"])
def delete_user(user_id):
    return jsonify({"message": f"User {user_id} deleted"}), 204

print("Flask routes defined")
````

## Complex: Query Parameters & Error Handling

````python
from flask import Flask, jsonify, request

app = Flask(__name__)

@app.route("/search")
def search():
    query = request.args.get("q", "")
    page = request.args.get("page", 1, type=int)
    limit = request.args.get("limit", 10, type=int)
    return jsonify({"query": query, "page": page, "limit": limit})

@app.route("/validate", methods=["POST"])
def validate_data():
    data = request.json
    if not data or "name" not in data:
        return jsonify({"error": "Missing 'name' field"}), 400
    if len(data["name"]) < 1:
        return jsonify({"error": "Name too short"}), 422
    return jsonify({"status": "valid", "data": data}), 200

@app.errorhandler(404)
def not_found(error):
    return jsonify({"error": "Not found"}), 404

@app.errorhandler(500)
def server_error(error):
    return jsonify({"error": "Internal server error"}), 500

print("Flask app with validation defined")
````

## Testing with Werkzeug Test Client

````python
from flask import Flask, jsonify, request

app = Flask(__name__)

@app.route("/items/<int:item_id>")
def get_item(item_id):
    items = {1: "Widget", 2: "Gadget", 3: "Gizmo"}
    if item_id not in items:
        return jsonify({"error": "Item not found"}), 404
    return jsonify({"id": item_id, "name": items[item_id]})

client = app.test_client()

response = client.get("/items/1")
print("GET /items/1:", response.get_json())

response = client.get("/items/999")
print("GET /items/999 status:", response.status_code)
````

## What is UV?
**UV** is a modern, fast Python package manager and script runner. It replaces `pip` and `venv`:
- Faster than pip (written in Rust)
- No need to activate venv: `uv run script.py` runs directly
- Install packages: `uv pip install package` or `uv run --with package script.py`
- Handles venv creation automatically

Traditional Python way: `python -m venv venv && source venv/bin/activate && pip install flask`

UV way: `uv run --with flask python app.py` (no activation needed)

## How to Run & Call Flask APIs

### Setup: Using Python or UV
```bash
# With traditional Python + pip:
python -m venv venv
source venv/bin/activate  # Windows: venv\Scripts\activate
pip install flask

# With UV (faster, simpler):
uv venv
source .venv/bin/activate  # Windows: .venv\Scripts\activate
uv pip install flask
# OR direct execution: uv run --with flask python app.py
```

### Run Flask Server (Default & Custom Ports)
```bash
# Default port 5000:
python app.py
uv run app.py

# Custom port 8080 (modify app.py):
# app.run(host='0.0.0.0', port=8080, debug=True)
python app.py
uv run app.py

# Or pass via command-line:
python app.py -- --port 8080
uv run app.py -- --port 8080

# Custom port via environment variable:
export FLASK_PORT=9000  # Windows: set FLASK_PORT=9000
python app.py
FLASK_PORT=9000 uv run app.py
```

### Custom Port in Python Code
```python
from flask import Flask

app = Flask(__name__)

@app.route('/')
def home():
    return 'Hello'

if __name__ == '__main__':
    # Default port 5000
    app.run(debug=True)
    
    # Custom port 8080
    app.run(host='0.0.0.0', port=8080, debug=True)
    
    # Custom port from environment variable
    import os
    port = int(os.getenv('FLASK_PORT', 5000))
    app.run(host='0.0.0.0', port=port, debug=True)
```

### Call Flask API from Python
```python
import requests

BASE_URL = 'http://localhost:5000'  # Change port as needed

# GET
response = requests.get(f'{BASE_URL}/users/1')
print(response.json())

# POST
payload = {"name": "Alice", "email": "alice@example.com"}
response = requests.post(f'{BASE_URL}/create', json=payload)
print(response.json())

# PUT
updates = {"name": "Alice Updated"}
response = requests.put(f'{BASE_URL}/users/1', json=updates)

# DELETE
response = requests.delete(f'{BASE_URL}/users/1')
print(response.status_code)
```

- `204` No Content
- `400` Bad Request
- `404` Not Found
- `500` Internal Server Error
