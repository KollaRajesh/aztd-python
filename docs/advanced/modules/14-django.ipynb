Django requires a full project structure to run. Code in this notebook is **reference documentation only** (not executable in Jupyter) because:

1. Django needs `manage.py` - a project management script
2. Models must be in a Django app with proper config
3. Database migrations are required before queries work
4. Views need URL routing and WSGI/ASGI server
5. Settings and environment setup are mandatory

The patterns shown below are real Django code - copy them into your project's files (models.py, views.py, urls.py, etc.). See **Setup** section at the bottom for how to create and run a Django project.

## Setup: Using Python or UV
```bash
# With traditional Python + pip:
python -m venv venv
source venv/bin/activate  # Windows: venv\Scripts\activate
pip install django

# With UV (faster, simpler):
uv venv
source .venv/bin/activate  # Windows: .venv\Scripts\activate
uv pip install django
# OR direct: uv run --with django django-admin startproject myproject
```

## Create a Django Project
```bash
# Create project structure
django-admin startproject myproject
cd myproject

# Create an app (app is a module within project)
python manage.py startapp myapp

# Run migrations (setup database)
python manage.py migrate

# Start development server
python manage.py runserver  # Runs on http://localhost:8000

# Create admin user
python manage.py createsuperuser  # Then visit http://localhost:8000/admin/
```

## File Structure
```
myproject/
  manage.py              # Project management script

    settings.py         # Configuration
    urls.py            # URL routing
  myapp/
    models.py          # Database schema
    views.py           # Request handlers
    urls.py            # App URL patterns
```

## Key Differences: Django vs Flask vs FastAPI
| Feature | Django | Flask | FastAPI |
|---------|--------|-------|----------|
| Built-in ORM | Yes (complex) | No | No |
| Admin Interface | Yes (auto) | No | No |
| Database Migrations | Yes | No | No |
| Authentication | Yes | No | No |
| Async Support | Limited | No | Full |
| Learning Curve | Steep | Gentle | Moderate |
| Best For | Large projects | Simple APIs | Modern async APIs |

See [Django Documentation](https://docs.djangoproject.com/) for full details.
