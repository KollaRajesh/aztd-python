## Simple: Arrays & Basic Operations

```python
import numpy as np

# Create arrays
arr = np.array([1, 2, 3, 4, 5])
zeros = np.zeros(5)
range_arr = np.arange(0, 10, 2)

print("Array:", arr)
print("Zeros:", zeros)
print("Range:", range_arr)
```

```python
# Operations
print("arr * 2:", arr * 2)
print("arr + arr:", arr + arr)
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Std:", np.std(arr))
```

## Medium: Matrices & Linear Algebra

```python
import numpy as np

# 2D arrays
matrix = np.array([[1, 2], [3, 4]])
print("Matrix shape:", matrix.shape)
print("Matrix:\n", matrix)
```

```python
# Matrix operations
matrix2 = np.array([[5, 6], [7, 8]])
result = np.dot(matrix, matrix2)
print("Matrix multiplication:\n", result)
```

```python
# Statistics
data = np.array([1, 2, 3, 4, 5])
print("Std:", np.std(data))
print("75th percentile:", np.percentile(data, 75))
```

## Complex: Broadcasting & Vectorization

```python
import numpy as np

# Broadcasting
a = np.array([1, 2, 3])
b = np.array([[1], [2], [3]])
result = a + b
print("Broadcasting result shape:", result.shape)
print(result)
```

```python
# Vectorized operations
x = np.linspace(0, 2*np.pi, 10)
y = np.sin(x)
print("Sin values:", y)
```

```python
# Advanced indexing
arr = np.arange(20).reshape(4, 5)
mask = arr > 10
filtered = arr[mask]
print("Values > 10:", filtered)
```
