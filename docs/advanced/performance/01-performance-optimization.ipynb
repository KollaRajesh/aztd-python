# Performance Optimization

Techniques for profiling, benchmarking, and optimizing Python code.

## 1. Timing: Measure Execution Time

```python
import time

def slow_function():
    """Intentionally slow function"""
    total = 0
    for i in range(1000000):
        total += i
    return total

# Time it
start = time.perf_counter()
result = slow_function()
elapsed = time.perf_counter() - start

print(f"Result: {result}")
print(f"Time: {elapsed:.4f} seconds")
```

## 2. Optimization: Caching with @functools.lru_cache

```python
from functools import lru_cache
import time

# Without cache (slow)
def fib_slow(n):
    if n <= 1:
        return n
    return fib_slow(n-1) + fib_slow(n-2)

# With cache (fast)
@lru_cache(maxsize=None)
def fib_fast(n):
    if n <= 1:
        return n
    return fib_fast(n-1) + fib_fast(n-2)

# Compare
n = 30

start = time.perf_counter()
result_slow = fib_slow(n)
time_slow = time.perf_counter() - start

start = time.perf_counter()
result_fast = fib_fast(n)
time_fast = time.perf_counter() - start

print(f"Slow: {time_slow:.4f}s")
print(f"Fast (cached): {time_fast:.4f}s")
print(f"Speedup: {time_slow/time_fast:.0f}x faster")
```

## 3. List vs Set Lookup Performance

```python
import time

# Create large collections
numbers = list(range(100000))
number_set = set(numbers)

# Lookup in list (O(n) - slow)
start = time.perf_counter()
for _ in range(1000):
    _ = 99999 in numbers
time_list = time.perf_counter() - start

# Lookup in set (O(1) - fast)
start = time.perf_counter()
for _ in range(1000):
    _ = 99999 in number_set
time_set = time.perf_counter() - start

print(f"List lookup (1000x): {time_list:.4f}s")
print(f"Set lookup (1000x): {time_set:.4f}s")
print(f"Set is {time_list/time_set:.0f}x faster")
```

## 4. List Comprehension vs Loop

```python
import time

# Loop approach
start = time.perf_counter()
result_loop = []
for i in range(100000):
    if i % 2 == 0:
        result_loop.append(i * 2)
time_loop = time.perf_counter() - start

# List comprehension
start = time.perf_counter()
result_comp = [i * 2 for i in range(100000) if i % 2 == 0]
time_comp = time.perf_counter() - start

print(f"Loop approach: {time_loop:.4f}s")
print(f"List comprehension: {time_comp:.4f}s")
print(f"Comprehension is {time_loop/time_comp:.1f}x faster")
```

## 5. String Concatenation: += vs join()

```python
import time

# Concatenation with +=
start = time.perf_counter()
result = ""
for i in range(10000):
    result += f"line {i}\n"
time_concat = time.perf_counter() - start

# Using join()
start = time.perf_counter()
lines = [f"line {i}" for i in range(10000)]
result = "\n".join(lines)
time_join = time.perf_counter() - start

print(f"Concatenation (+=): {time_concat:.4f}s")
print(f"join(): {time_join:.4f}s")
print(f"join() is {time_concat/time_join:.0f}x faster")
```

## 6. NumPy vs Pure Python

```python
import time
import numpy as np

# Pure Python
start = time.perf_counter()
result_py = [x ** 2 for x in range(1000000)]
time_py = time.perf_counter() - start

# NumPy
start = time.perf_counter()
arr = np.arange(1000000)
result_np = arr ** 2
time_np = time.perf_counter() - start

print(f"Pure Python: {time_py:.4f}s")
print(f"NumPy: {time_np:.4f}s")
print(f"NumPy is {time_py/time_np:.0f}x faster")
```

## 7. Memory Profiling

```python
# Example: Memory usage of different data structures
import sys

list_data = list(range(10000))
tuple_data = tuple(range(10000))
set_data = set(range(10000))

print(f"List size: {sys.getsizeof(list_data)} bytes")
print(f"Tuple size: {sys.getsizeof(tuple_data)} bytes")
print(f"Set size: {sys.getsizeof(set_data)} bytes")
```

## 8. Generator Expressions (Memory Efficient)

```python
import sys

# List (stores everything in memory)
list_comp = [x ** 2 for x in range(1000000)]
print(f"List comprehension size: {sys.getsizeof(list_comp) / 1024 / 1024:.2f} MB")

# Generator (lazy evaluation - minimal memory)
gen_expr = (x ** 2 for x in range(1000000))
print(f"Generator expression size: {sys.getsizeof(gen_expr)} bytes")

# Use generator
first_10 = [next(gen_expr) for _ in range(10)]
print(f"First 10 squares: {first_10}")
```
