## 2.20 HTTP Requests

```python
import requests

# GET request
response = requests.get('https://jsonplaceholder.typicode.com/posts/1')
print(f"Status: {response.status_code}")
print(f"Data: {response.json()}")
```

```python
import requests

# POST request
data = {"title": "My Post", "body": "Content", "userId": 1}
response = requests.post('https://jsonplaceholder.typicode.com/posts', json=data)
print(f"Status: {response.status_code}")
print(f"Created: {response.json()}")
```

## 2.21 Async HTTP

```python
import asyncio
import aiohttp

async def fetch(session, url):
    async with session.get(url) as response:
        return await response.json()

async def fetch_multiple(urls):
    async with aiohttp.ClientSession() as session:
        tasks = [fetch(session, url) for url in urls]
        return await asyncio.gather(*tasks)

# URLs to fetch
urls = [
    'https://jsonplaceholder.typicode.com/posts/1',
    'https://jsonplaceholder.typicode.com/posts/2',
    'https://jsonplaceholder.typicode.com/posts/3'
]

# Run async fetch
results = await fetch_multiple(urls)
print(f"Fetched {len(results)} posts")
```

## 2.22 DNS Lookup

```python
import socket

# DNS lookup
hostname = "google.com"
ip_address = socket.gethostbyname(hostname)
print(f"{hostname} -> {ip_address}")

# Reverse DNS
reverse = socket.gethostbyaddr(ip_address)
print(f"{ip_address} -> {reverse[0]}")
```

## 2.23 Port Scanning

```python
import socket

def is_port_open(host, port):
    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    result = sock.connect_ex((host, port))
    sock.close()
    return result == 0

# Scan common ports
common_ports = [80, 443, 8080, 3306, 5432]

for port in common_ports:
    status = "open" if is_port_open('localhost', port) else "closed"
    print(f"Port {port}: {status}")
```
